```python
import os
import pandas as pd
import numpy as np

np.random.seed(42)
n = 50

ids_productos = [f"PROD_{i:03d}" for i in range(1, n + 1)]
nombres = [f"Artículo Local {i}" for i in range(1, n + 1)]
stock = np.random.randint(10, 300, size=n)
precio = np.round(np.random.uniform(10.0, 250.0, size=n), 2)
fechas = pd.date_range(start="2026-01-01", periods=n, freq="D")
categorias = np.random.choice(["Abarrotes", "Papelería", "Bebidas", "Snacks"], size=n)
activo = np.random.choice([True, False], size=n, p=[0.85, 0.15])

df_muestra = pd.DataFrame({
    "producto_id": ids_productos,
    "nombre_producto": nombres,
    "stock": stock,
    "precio": precio,
    "ultima_actualizacion": fechas,
    "categoria": categorias,
    "activo": activo
})

df_muestra["categoria"] = df_muestra["categoria"].astype("category")

print("--- Primeras filas del dataset de muestra ---")
display(df_muestra.head())

print("\n--- Tipos de datos originales ---")
print(df_muestra.dtypes)

os.makedirs("datos", exist_ok=True)
csv_path = "datos/muestra_tienda.csv"
parquet_path = "datos/muestra_tienda.parquet"

df_muestra.to_csv(csv_path, index=False)
df_muestra.to_parquet(parquet_path, index=False)

df_leido_parquet = pd.read_parquet(parquet_path)
print("\n--- Tipos de datos recuperados desde Parquet ---")
print(df_leido_parquet.dtypes)

tamanio_csv = os.path.getsize(csv_path)
tamanio_parquet = os.path.getsize(parquet_path)

print(f"\nTamaño en disco (CSV): {tamanio_csv} bytes")
print(f"Tamaño en disco (Parquet): {tamanio_parquet} bytes")
print(f"Parquet es {tamanio_csv / tamanio_parquet:.2f} veces más compacto que el CSV.")
```

¿Qué hicimos y por qué?
Para esta práctica creamos una tabla de prueba con 50 registros que incluía 6 tipos de datos diferentes (texto, números enteros, decimales, fechas, categorías y booleanos). La idea principal era comparar cómo se comportan dos formatos muy conocidos al guardarlos en la computadora: el clásico CSV y el formato avanzado Apache Parquet.

¿Qué ventajas notamos al usar Parquet?
No se pierden los tipos de datos: Nos dimos cuenta de que cuando guardas un archivo en CSV, al volver a abrirlo pandas a veces ya no sabe si una columna era fecha o categoría y te lo convierte todo en texto plano.

¿En qué caso NO nos convendría usar Parquet?
Un detalle importante es que Parquet no se recomendaría si necesitamos abrir el archivo rápidamente para revisarlo a simple vista en un bloc de notas

## Definición de funciones reutilizables de limpieza

```python
# Función para normalizar columnas de texto.
def limpiar_texto(serie):

    # Convertimos los valores a texto conservando los faltantes.
    serie = serie.astype("string")

    # Eliminamos espacios al inicio y al final.
    serie = serie.str.strip()

    # Convertimos el texto a mayúsculas.
    serie = serie.str.upper()

    # Devolvemos la columna normalizada.
    return serie


# Función para convertir valores de texto a números.
def a_numero(serie):

    # Convertimos los valores a texto conservando los faltantes.
    serie = serie.astype("string")

    # Eliminamos el símbolo monetario.
    serie = serie.str.replace("$", "", regex=False)

    # Eliminamos la abreviatura de piezas.
    serie = serie.str.replace("pz", "", regex=False)

    # Eliminamos espacios sobrantes.
    serie = serie.str.strip()

    # Convertimos a valores numéricos.
    # Los valores que no se puedan convertir quedarán como NaN.
    return pd.to_numeric(serie, errors="coerce")
```

## Preparamos una copia para realizar el diagnóstico
## Preparar copia sin duplicados

```python
# Creamos una copia sin duplicados a partir del dataset original.
ventas_sin_duplicados = ventas_originales.drop_duplicates().copy()

# Contamos las filas que necesitan normalización,
# después de eliminar los duplicados.
filas_sucursal_limpieza = (
    ventas_sin_duplicados["sucursal_id"].astype("string")
    != limpiar_texto(ventas_sin_duplicados["sucursal_id"])
).sum()

# Mostramos los resultados.
print("Filas originales:", len(ventas))

print("Filas después de quitar duplicados:",
      len(ventas_sin_duplicados))

print("Filas que necesitan normalización de sucursal:",
      filas_sucursal_limpieza)
```

```python
# Mostramos los códigos de sucursal antes de limpiarlos.
print("SUCURSALES ANTES DE LA LIMPIEZA")

# value_counts() cuenta cuántas veces aparece cada valor.
# dropna=False incluye los valores faltantes, si existen.
display(
    ventas["sucursal_id"].value_counts(dropna=False)
)

# Contamos cuántos códigos distintos existen originalmente.
print(
    "Códigos de sucursal originales:",
    ventas["sucursal_id"].nunique()
)

# Aplicamos la función de normalización a una copia temporal.
# Todavía no modificamos el DataFrame original.
sucursales_normalizadas = limpiar_texto(
    ventas["sucursal_id"]
)

# Contamos cuántos códigos diferentes quedarían.
print(
    "Códigos después de normalizar:",
    sucursales_normalizadas.nunique()
)

# Identificamos cuántas filas cambiarían con la normalización.
filas_sucursal_afectadas = (
    ventas["sucursal_id"].astype("string")
    != sucursales_normalizadas
).sum()

print(
    "Filas que requieren normalización:",
    filas_sucursal_afectadas
)

# Mostramos los códigos normalizados y sus frecuencias.
print("SUCURSALES DESPUÉS DE NORMALIZAR")

display(
    sucursales_normalizadas.value_counts(dropna=False)
)
```

### Diagnóstico de las columnas cantidad y precio_unitario

Se identificarán los valores faltantes, los formatos
inconsistentes y los registros que no pueden convertirse
correctamente a valores numéricos.

El diagnóstico se realizará sobre los 600 registros
resultantes de excluir los duplicados, sin modificar
todavía el dataset original.

```python
# Seleccionamos las columnas numéricas que vamos a diagnosticar.
columnas_numericas = ["cantidad", "precio_unitario"]

# Recorremos cada columna para analizar sus problemas.
for columna in columnas_numericas:

    # Mostramos el nombre de la columna analizada.
    print(f"\nDIAGNÓSTICO DE: {columna}")

    # Obtenemos los valores originales, sin modificarlos.
    valores_originales = ventas_sin_duplicados[columna]

    # Contamos los valores faltantes originales.
    nulos_originales = valores_originales.isna().sum()

    # Convertimos temporalmente los valores a números.
    valores_convertidos = a_numero(valores_originales)

    # Contamos los valores faltantes después de convertir.
    nulos_convertidos = valores_convertidos.isna().sum()

    # Identificamos los valores no nulos que no pudieron convertirse.
    conversiones_fallidas = (
        valores_originales.notna()
        & valores_convertidos.isna()
    ).sum()

    # Identificamos valores que contienen símbolos o unidades.
    valores_con_formato = valores_originales.astype("string").str.contains(
        r"\$|pz",
        case=False,
        regex=True,
        na=False
    ).sum()

    # Mostramos los resultados del diagnóstico.
    print("Nulos originales:", nulos_originales)
    print("Nulos después de convertir:", nulos_convertidos)
    print("Conversiones fallidas:", conversiones_fallidas)
    print("Valores con símbolos o unidades:", valores_con_formato)

    # Mostramos ejemplos de valores que no pudieron convertirse.
    print("Ejemplos de conversiones fallidas:")

    display(
        valores_originales[
            valores_originales.notna()
            & valores_convertidos.isna()
        ].head(10)
    )
```

### Diagnóstico de la columna fecha

Se identificarán los formatos de fecha presentes en el dataset
y se comprobará si existen registros que no pueden convertirse
correctamente al tipo datetime.

El diagnóstico se realiza sobre los 600 registros sin duplicados.

```python
# Obtenemos las fechas originales sin modificar el DataFrame.
fechas_originales = ventas_sin_duplicados["fecha"]

# Intentamos convertir las fechas con formato año-mes-día.
fechas_iso = pd.to_datetime(
    fechas_originales,
    format="%Y-%m-%d",
    errors="coerce"
)

# Intentamos convertir las fechas con formato día/mes/año.
fechas_dmy = pd.to_datetime(
    fechas_originales,
    format="%d/%m/%Y",
    errors="coerce"
)

# Combinamos ambas conversiones.
# Si la primera produjo NaT, usamos la segunda.
fechas_convertidas = fechas_iso.fillna(fechas_dmy)

# Contamos cuántas fechas pertenecen a cada formato.
cantidad_iso = fechas_iso.notna().sum()
cantidad_dmy = fechas_dmy.notna().sum()

# Identificamos fechas presentes que no pudieron convertirse.
fechas_invalidas = (
    fechas_originales.notna()
    & fechas_convertidas.isna()
).sum()

# Contamos valores faltantes originales.
fechas_nulas = fechas_originales.isna().sum()

# Mostramos los resultados.
print("DIAGNÓSTICO DE FECHAS")

print("Fechas en formato AAAA-MM-DD:", cantidad_iso)
print("Fechas en formato DD/MM/AAAA:", cantidad_dmy)
print("Fechas nulas originales:", fechas_nulas)
print("Fechas que no pudieron convertirse:", fechas_invalidas)

# Mostramos ejemplos de fechas que no pudieron convertirse.
print("\nEJEMPLOS DE FECHAS INVÁLIDAS")

display(
    fechas_originales[
        fechas_originales.notna()
        & fechas_convertidas.isna()
    ].head(10)
)
```

## Limpieza definitiva y verificaciones

## Aplicación de la limpieza

```python
# ============================================================
# TAREA 3 - LIMPIEZA DEFINITIVA DEL DATASET DE VENTAS
# ============================================================

# Creamos una copia independiente de los datos sin duplicados.
# Así conservamos intactos los DataFrames utilizados en los diagnósticos.
ventas_limpias = ventas_sin_duplicados.copy()


# ------------------------------------------------------------
# 1. NORMALIZACIÓN DE SUCURSALES
# ------------------------------------------------------------

# Eliminamos espacios sobrantes y convertimos los códigos a mayúsculas.
ventas_limpias["sucursal_id"] = limpiar_texto(
    ventas_limpias["sucursal_id"]
)


# ------------------------------------------------------------
# 2. CONVERSIÓN DE COLUMNAS NUMÉRICAS
# ------------------------------------------------------------

# Eliminamos símbolos y convertimos los precios a números.
ventas_limpias["precio_unitario"] = a_numero(
    ventas_limpias["precio_unitario"]
)

# Eliminamos unidades de texto y convertimos las cantidades a números.
ventas_limpias["cantidad"] = a_numero(
    ventas_limpias["cantidad"]
)


# ------------------------------------------------------------
# 3. CONVERSIÓN DE FECHAS
# ------------------------------------------------------------

# Convertimos las fechas usando los dos formatos detectados
# durante el diagnóstico: AAAA-MM-DD y DD/MM/AAAA.

fechas_iso_limpieza = pd.to_datetime(
    ventas_limpias["fecha"],
    format="%Y-%m-%d",
    errors="coerce"
)

fechas_dmy_limpieza = pd.to_datetime(
    ventas_limpias["fecha"],
    format="%d/%m/%Y",
    errors="coerce"
)

# Conservamos las fechas ISO válidas y completamos con las DMY.
ventas_limpias["fecha"] = fechas_iso_limpieza.fillna(
    fechas_dmy_limpieza
)


# ------------------------------------------------------------
# 4. TRATAMIENTO DE VALORES FALTANTES
# ------------------------------------------------------------

# Contamos los valores faltantes antes de imputar.
nulos_cantidad = ventas_limpias["cantidad"].isna().sum()
nulos_precio = ventas_limpias["precio_unitario"].isna().sum()


# 4.1 IMPUTACIÓN DE CANTIDAD POR PRODUCTO

# Calculamos la mediana de cantidad correspondiente a cada producto.
mediana_cantidad_producto = ventas_limpias.groupby(
    "producto_id"
)["cantidad"].transform("median")

# Redondeamos las medianas para mantener cantidades enteras.
mediana_cantidad_producto = mediana_cantidad_producto.round()

# Identificamos los casos que no pueden imputarse por producto.
sin_mediana_cantidad = (
    ventas_limpias["cantidad"].isna()
    & mediana_cantidad_producto.isna()
).sum()

# Imputamos las cantidades faltantes con la mediana de su producto.
ventas_limpias["cantidad"] = ventas_limpias["cantidad"].fillna(
    mediana_cantidad_producto
)

# Calculamos una mediana general de respaldo.
mediana_general_cantidad = round(
    ventas_limpias["cantidad"].median()
)

# Aplicamos el respaldo únicamente si quedan valores faltantes.
ventas_limpias["cantidad"] = ventas_limpias["cantidad"].fillna(
    mediana_general_cantidad
)


# 4.2 IMPUTACIÓN DE PRECIO UNITARIO POR PRODUCTO

# Calculamos la mediana del precio correspondiente a cada producto.
mediana_precio_producto = ventas_limpias.groupby(
    "producto_id"
)["precio_unitario"].transform("median")

# Identificamos los casos que no pueden imputarse por producto.
sin_mediana_precio = (
    ventas_limpias["precio_unitario"].isna()
    & mediana_precio_producto.isna()
).sum()

# Imputamos los precios faltantes con la mediana de su producto.
ventas_limpias["precio_unitario"] = ventas_limpias[
    "precio_unitario"
].fillna(mediana_precio_producto)

# Calculamos una mediana general de respaldo.
mediana_general_precio = ventas_limpias[
    "precio_unitario"
].median()

# Aplicamos el respaldo únicamente si quedan valores faltantes.
ventas_limpias["precio_unitario"] = ventas_limpias[
    "precio_unitario"
].fillna(mediana_general_precio)


# ------------------------------------------------------------
# 5. COMPROBACIÓN INICIAL DE LA LIMPIEZA
# ------------------------------------------------------------

print("RESUMEN DE LIMPIEZA")
print("Registros originales:", len(ventas_originales))
print("Registros limpios:", len(ventas_limpias))
print("Duplicados restantes:", ventas_limpias.duplicated().sum())

print("\nRESUMEN DE IMPUTACIÓN")
print("Cantidades faltantes originales:", nulos_cantidad)
print("Precios faltantes originales:", nulos_precio)

print("Cantidades imputadas con respaldo general:",
      sin_mediana_cantidad)

print("Precios imputados con respaldo general:",
      sin_mediana_precio)

print("\nNULOS DESPUÉS DE LIMPIAR")
print(ventas_limpias.isnull().sum())

print("\nTIPOS DE DATOS")
print(ventas_limpias.dtypes)

print("\nSUCURSALES NORMALIZADAS")
print(ventas_limpias["sucursal_id"].value_counts())

print("\nPRIMERAS FILAS DEL DATASET LIMPIO")
display(ventas_limpias.head())
```

## Verificar que no existan cantidades fraccionarias, cantidades no positivas o precios no positivos.

```python
# ============================================================
# TAREA 3 - VALIDACIÓN DE VALORES LIMPIOS
# ============================================================

# Identificamos cantidades que tienen parte decimal.
cantidades_fraccionarias = (
    ventas_limpias["cantidad"] % 1 != 0
).sum()

# Identificamos cantidades iguales o menores que cero.
cantidades_no_positivas = (
    ventas_limpias["cantidad"] <= 0
).sum()

# Identificamos precios iguales o menores que cero.
precios_no_positivos = (
    ventas_limpias["precio_unitario"] <= 0
).sum()

# Identificamos fechas que no pudieron convertirse.
fechas_invalidas_finales = (
    ventas_limpias["fecha"].isna().sum()
)

# Mostramos los resultados.
print("VALIDACIÓN DE VALORES LIMPIOS")

print("Cantidades fraccionarias:", cantidades_fraccionarias)
print("Cantidades no positivas:", cantidades_no_positivas)
print("Precios no positivos:", precios_no_positivos)
print("Fechas inválidas:", fechas_invalidas_finales)

# Comprobamos que los datos cumplen estas condiciones.
assert cantidades_fraccionarias == 0, "Existen cantidades fraccionarias"
assert cantidades_no_positivas == 0, "Existen cantidades no positivas"
assert precios_no_positivos == 0, "Existen precios no positivos"
assert fechas_invalidas_finales == 0, "Existen fechas inválidas"

print("\nTodas las validaciones de valores fueron superadas.")
```

### Bitácora de limpieza de datos

Se documentaron los problemas de calidad identificados en el conjunto de datos de ventas, incluyendo registros duplicados, inconsistencias en los códigos de sucursal, formatos numéricos, valores faltantes y formatos de fecha.

## Exportar y verificar el archivo Parquet

```python
# ============================================================
# TAREA 3 - EXPORTACIÓN Y VERIFICACIÓN DEL ARCHIVO PARQUET
# ============================================================

# Importamos Path para trabajar con carpetas y archivos.
from pathlib import Path

# Definimos la carpeta donde guardaremos los resultados.
carpeta_resultados = Path("resultados")

# Creamos la carpeta si todavía no existe.
carpeta_resultados.mkdir(parents=True, exist_ok=True)

# Definimos la ruta completa del archivo de salida.
ruta_parquet = carpeta_resultados / "ventas_limpias.parquet"

# Guardamos el DataFrame limpio en formato Parquet.
# index=False evita guardar el índice de pandas como columna adicional.
ventas_limpias.to_parquet(ruta_parquet, index=False)

# Comprobamos que el archivo exista físicamente.
assert ruta_parquet.is_file(), "No se generó el archivo Parquet."

print("Archivo generado correctamente:", ruta_parquet)

# Volvemos a leer el archivo para verificar su contenido.
ventas_verificacion = pd.read_parquet(ruta_parquet)

# Comprobamos que tenga las mismas dimensiones.
assert ventas_verificacion.shape == ventas_limpias.shape, \
    "Las dimensiones del archivo no coinciden."

# Comprobamos que los datos guardados sean iguales a los originales limpios.
pd.testing.assert_frame_equal(
    ventas_verificacion.reset_index(drop=True),
    ventas_limpias.reset_index(drop=True)
)

print("\nVERIFICACIÓN DEL ARCHIVO PARQUET")
print("Dimensiones:", ventas_verificacion.shape)
print("Duplicados:", ventas_verificacion.duplicated().sum())
print("Valores nulos:", ventas_verificacion.isna().sum().sum())

print("\nEl archivo Parquet se guardó y verificó correctamente.")
```
